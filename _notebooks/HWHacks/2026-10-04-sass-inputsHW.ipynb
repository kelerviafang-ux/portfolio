## Refined Signup Inputs

The signup fields use only the [OCS input grammar](https://pages.opencodingsociety.com/sass/inputs/): `ocs__input large gradient` for Assistance Category, `ocs__input small` for Volunteer First Name, and `ocs__input medium` for Volunteer Email. The shared theme supplies size, color, and focus styling; no inline styles or custom classes are needed.

Edit the HTML in the **UI Runner**, then press **Run** or **Ctrl+Enter / Cmd+Enter**. Type into each field and use Tab to check keyboard focus. **Save** keeps edits in this browser; **Clear Storage** restores the submitted HTML.

```python
%%html
<!-- UI_RUNNER: Signup inputs using OCS size and style modifiers | editor: html -->
<input type="text" class="ocs__input large gradient" placeholder="Assistance Category" aria-label="Assistance Category">
<input type="text" class="ocs__input small" placeholder="Volunteer First Name" aria-label="Volunteer First Name">
<input type="email" class="ocs__input medium" placeholder="Volunteer Email" aria-label="Volunteer Email">
```
