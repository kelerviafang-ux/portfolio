## SASS Grid Homework

Edit the JavaScript array of HTML strings below, then press **Run** or **Ctrl+Enter / Cmd+Enter**. The UI Runner supplies `outputElement`; a Python kernel is not required. Changes render only when you run the code.

The example contains a three-column temperature grid and a four-column calculator grid. A header spans the temperature grid; accent cells mark Beaker 3 and the Avg key. The calculator keys are layout examples.

Use **Save** to keep your edits in this browser. **Clear Storage** restores the starter; press Run to render it again. The checks show **PASS** or **FIX** for each requirement. Try removing a grid class or adding an inline style, then run again to test the checks. Resize the page and move the pointer over the cells to inspect the shared layout and hover styles.

// Edit the HTML strings, then press Run.
// outputElement is provided by the UI Runner.
let myHTML = [
  '<div class="ocs__grid ocs__grid--standard cols-3">',
  '  <div class="ocs__grid-cell ocs__grid-cell--header">Temperature Trial (°C)</div>',
  '  <div class="ocs__grid-cell">Beaker 1: 22.4</div>',
  '  <div class="ocs__grid-cell">Beaker 2: 22.9</div>',
  '  <div class="ocs__grid-cell ocs__grid-cell--accent">Beaker 3: 31.6</div>',
  '</div>',
  '<div class="ocs__grid ocs__grid--calculator">',
  '  <div class="ocs__grid-cell">7</div>',
  '  <div class="ocs__grid-cell">8</div>',
  '  <div class="ocs__grid-cell">9</div>',
  '  <div class="ocs__grid-cell ocs__grid-cell--accent">Avg</div>',
  '</div>'
].join('');

// ---- You do not need to change anything below this line ----
// Parse the edited markup so each check examines the correct grid.
const markup = document.createElement('template');
markup.innerHTML = myHTML;
const elements = [...markup.content.querySelectorAll('*')];
const allowedClasses = new Set([
  'ocs__grid', 'ocs__grid--standard', 'ocs__grid--calculator',
  'ocs__grid-cell', 'ocs__grid-cell--header', 'ocs__grid-cell--accent',
  'ocs__grid-cell--muted', 'cols-2', 'cols-3'
]);
const madeUp = [...new Set(elements.flatMap(element => [...element.classList]))]
  .filter(className => !allowedClasses.has(className));
const grids = [...markup.content.children];
const standard = grids[0];
const calculator = grids[1];
const standardCells = standard ? [...standard.children] : [];
const calculatorCells = calculator ? [...calculator.children] : [];
const allCells = [...standardCells, ...calculatorCells];
const hasClass = (element, className) => Boolean(element?.classList.contains(className));
const results = [
  ['No inline style attributes', !elements.some(element => element.hasAttribute('style'))],
  ['No made up classes' + (madeUp.length ? ' (found: ' + madeUp.join(', ') + ')' : ''), madeUp.length === 0],
  ['Both roots use ocs__grid', grids.length === 2 && grids.every(grid => hasClass(grid, 'ocs__grid'))],
  ['Grid 1 uses the standard variant with three columns', hasClass(standard, 'ocs__grid--standard') && hasClass(standard, 'cols-3')],
  ['Grid 2 uses the calculator variant', hasClass(calculator, 'ocs__grid--calculator')],
  ['Each grid has four boxes, all using ocs__grid-cell', standardCells.length === 4 && calculatorCells.length === 4 && allCells.every(cell => hasClass(cell, 'ocs__grid-cell'))],
  ['The temperature title is a header cell', hasClass(standardCells[0], 'ocs__grid-cell--header')],
  ['The outlier beaker and Avg key are accent cells', hasClass(standardCells[3], 'ocs__grid-cell--accent') && hasClass(calculatorCells[3], 'ocs__grid-cell--accent')]
];

// The shared styles apply inside one OCS container.
outputElement.innerHTML = '<div class="ocs__container"><div class="ocs__card">' +
  '<h3 class="ocs__section-title">Grid Checks</h3><ul></ul>' +
  myHTML + '</div></div>';
const checks = outputElement.querySelector('ul');
for (const [label, ok] of results) {
  const item = document.createElement('li');
  item.textContent = (ok ? 'PASS  ' : 'FIX  ') + label;
  checks.appendChild(item);
}
