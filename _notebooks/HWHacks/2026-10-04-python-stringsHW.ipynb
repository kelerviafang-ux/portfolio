## Crack the Final Transmission

This transmission has four comma-separated fields: date, agent codename, message code, and confidence. Press **Run** or **Ctrl+Enter / Cmd+Enter** to decode it in the browser.

The program strips outer whitespace, splits the row, and extracts each field by its zero-based index. It uses a slice to check the year and string methods to clean and search the message. Finally, f-strings rebuild the data as a Markdown table row and a JSON line with a numeric confidence and a lowercase Boolean.

The starter should print `True` for both checks, decode the message as `THE EAGLE LANDS AT DAWN`, and show `87%` in the Markdown row. Try changing the year or replacing `DAWN` with `NOON`, then run again. The first run downloads the browser Python runtime.

```python
# CODE_RUNNER: Homework - Crack the Final Transmission using slicing, methods, and formatting | python_runtime: browser, literal: true

raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

# strip() cleans the outside of the row; split(",") makes a four-item list.
segments = raw_transmission.strip().split(",")
print("Segments:", segments)

# List indexes start at zero, so the fourth field is at index 3.
date = segments[0]
agent = segments[1]
message_code = segments[2]
confidence_str = segments[3]
print("Date:", date, "| Agent:", agent, "| Message code:", message_code, "| Confidence:", confidence_str)

# The end index is excluded: [:4] takes exactly the first four characters.
year_check = date[:4] == "2026"
print("Year check:", year_check)

# replace() returns a new string with spaces in place of the message dashes.
message = message_code.replace("-", " ")

# lower() makes the search case-insensitive; IN returns a Boolean.
found_dawn = "dawn" in message.lower()
print("Found 'dawn':", found_dawn)

# An f-string inserts values, while the percent sign and table separators stay literal.
markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
print("Markdown row:", markdown_row)

# Double braces produce literal JSON braces inside an f-string.
# The confidence is unquoted, and str(...).lower() produces JSON's true/false.
json_line = (
    f'{{"date": "{date}", "agent": "{agent}", "message": "{message}", '
    f'"confidence": {confidence_str}, "contains_dawn": {str(found_dawn).lower()}}}'
)
print("JSON line:", json_line)
```
