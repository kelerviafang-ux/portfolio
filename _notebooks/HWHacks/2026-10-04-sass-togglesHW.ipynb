## Study Session Toggles

Press **Run** to create the panel. Each toggle has a different purpose: **Focus mode** hides distractions, **Study tips** reveals advice, and **Break reminders** shows a reminder. The **Active Toggles** section counts how many switches are on.

Edit the HTML and JavaScript in the **UI Runner**, then press **Run** or **Ctrl+Enter / Cmd+Enter**. Click a label to flip its switch, or use **Tab** and **Space**. Try all eight combinations and watch the count change from 0 to 3. **Save** keeps code edits in this browser; **Clear Storage** restores the starter. Running again resets all three switches to off.

```python
%%html
<!-- UI_RUNNER: Build a study session with three independent OCS toggles and an active count | editor: html -->
<div class="ocs__container" id="study-toggle-panel">
  <article class="ocs__card">
    <h2 class="ocs__section-title">Study Session Controls</h2>
    <p>Choose the tools that help you study. Every switch changes this panel.</p>

    <!-- A label wraps each native checkbox so clicking its text also flips it. -->
    <p>
      <label class="ocs__toggle">
        <input class="ocs__toggle-input" id="focus-toggle" type="checkbox">
        <span class="ocs__toggle-track" aria-hidden="true"></span>
        <span class="ocs__toggle-label">Focus mode</span>
      </label>
    </p>
    <p>
      <label class="ocs__toggle">
        <input class="ocs__toggle-input" id="tips-toggle" type="checkbox">
        <span class="ocs__toggle-track" aria-hidden="true"></span>
        <span class="ocs__toggle-label">Study tips</span>
      </label>
    </p>
    <p>
      <label class="ocs__toggle">
        <input class="ocs__toggle-input" id="break-toggle" type="checkbox">
        <span class="ocs__toggle-track" aria-hidden="true"></span>
        <span class="ocs__toggle-label">Break reminders</span>
      </label>
    </p>

    <section aria-labelledby="active-toggle-title">
      <h3 class="ocs__section-title" id="active-toggle-title">Active Toggles</h3>
      <p role="status" aria-live="polite"><strong id="active-toggle-count">0</strong> of 3 toggles are on.</p>
    </section>

    <p id="focus-status">Focus mode is off: distractions are visible.</p>
    <p id="tips-status">Study tips are off.</p>
    <p id="break-status">Break reminders are off.</p>

    <section class="ocs__callout" id="distraction-panel">
      <h3 class="ocs__section-title">Distractions</h3>
      <p>Game updates and social notifications can wait until after your study session.</p>
    </section>
    <section class="ocs__callout" id="study-tip-panel" hidden>
      <h3 class="ocs__section-title">Study Tip</h3>
      <p>Explain one concept in your own words, then test yourself without looking at your notes.</p>
    </section>
    <section class="ocs__callout" id="break-reminder-panel" hidden>
      <h3 class="ocs__section-title">Break Reminder</h3>
      <p>At your next stopping point, take a short break and stretch.</p>
    </section>
  </article>
</div>
<script>
(function () {
  // Scoped selectors keep this activity separate from other runners on the page.
  const root = typeof outputElement === "undefined"
    ? document.getElementById("study-toggle-panel")
    : outputElement.querySelector("#study-toggle-panel");
  const focusToggle = root.querySelector("#focus-toggle");
  const tipsToggle = root.querySelector("#tips-toggle");
  const breakToggle = root.querySelector("#break-toggle");
  const toggles = [focusToggle, tipsToggle, breakToggle];

  function updatePage() {
    // .checked is a Boolean: true means the switch is on, false means it is off.
    const focusOn = focusToggle.checked;
    const tipsOn = tipsToggle.checked;
    const breaksOn = breakToggle.checked;

    root.querySelector("#distraction-panel").hidden = focusOn;
    root.querySelector("#study-tip-panel").hidden = !tipsOn;
    root.querySelector("#break-reminder-panel").hidden = !breaksOn;
    root.querySelector("#focus-status").textContent = focusOn
      ? "Focus mode is on: distractions are hidden."
      : "Focus mode is off: distractions are visible.";
    root.querySelector("#tips-status").textContent = tipsOn
      ? "Study tips are on."
      : "Study tips are off.";
    root.querySelector("#break-status").textContent = breaksOn
      ? "Break reminders are on."
      : "Break reminders are off.";

    // filter keeps only checked switches; length gives a fresh count each time.
    root.querySelector("#active-toggle-count").textContent =
      toggles.filter(toggle => toggle.checked).length;
  }

  // change handles both pointer clicks and keyboard changes, unlike a click-only handler.
  toggles.forEach(toggle => toggle.addEventListener("change", updatePage));
  // Initialize once so the displayed content matches the switches before any interaction.
  updatePage();
})();
</script>
```
