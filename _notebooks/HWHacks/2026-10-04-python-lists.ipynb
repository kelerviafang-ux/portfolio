## Lists Homework - indexing, accumulator, count and filter on one dataset

```python
# CODE_RUNNER: Use indexing, an accumulator, counting, and filtering to summarize donations. | python_runtime: browser

donations_by_item = ["blankets", "socks", "blankets", "water", "socks", "blankets",
                     "meals", "socks", "blankets", "water", "blankets", "socks"]
costs = [2, 3, 2, 4, 3, 2, 1, 3, 2, 4, 2, 3]

# LESSON 1 - len() counts entries; index -1 selects the last entry.
print(len(donations_by_item))
print(donations_by_item[-1])

# LESSON 2 - Each loop iteration adds one cost to the accumulator.
total = 0
for cost in costs:
    total += cost
average = total / len(costs)
print("Total:", int(total))
print("Average:", int(average))

# LESSON 3 - get(item, 0) starts a new item's count at zero.
frequency = {}
for item in donations_by_item:
    frequency[item] = frequency.get(item, 0) + 1
print(frequency)

# Keep an item only when its donation count reaches the threshold of three.
frequent_items = []
for item, count in frequency.items():
    if count >= 3:
        frequent_items.append(item)
print(frequent_items)
```
