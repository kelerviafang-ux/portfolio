## JavaScript Runner

This example uses JavaScript to render the Poway Rotary Parade card, grid, table, and callout with the shared OCS SASS styles from the [Containers lesson](https://pages.opencodingsociety.com/navigation/sass/containers/lesson/). The container has a gray background; move the pointer over the card or a grid cell to see it highlight.

1. Edit JavaScript or the HTML strings in the editor below. Do not add `%%html` or `<script>` wrappers: the runner supplies `outputElement`.
2. Click **▶ Run** or press **Ctrl+Enter / Cmd+Enter** to render the changes immediately.
3. Changes appear only when you press **Run** or the keyboard shortcut. **■ Stop** clears the preview.
4. **💾 Save** stores your code in this browser; refresh to restore it. **🗑 Clear Storage** removes that saved version and restores the starter. **↻ Reset** clears the output; press Run to render again.
5. Change the title, location, start time, grid classes, or table rows; click the volunteer button to test a JavaScript event handler. Resize the window to check the responsive layout.

## Deployment

This notebook is the source for the published page. The existing build converts it to Markdown, then Jekyll renders the Liquid UI Runner include. The runner imports JavaScript using `site.baseurl`, so it works under `/portfolio` and on a root-domain site.

- GitHub Pages: commit this notebook and the runner changes, then push to `main`; the existing GitHub Actions workflow builds and deploys the site.
- A server that pulls this repository must also convert notebooks and rebuild Jekyll after `git pull` (or use the existing `make` notebook watcher). Pulling files alone does not rebuild a static site.
- Open `{{ '/hwhacks/sass-container-js-runner/' | relative_url }}` after deployment. Run executes JavaScript directly in the browser and needs no Python kernel or backend execution API.
- Browser edits and **Save** are local practice changes. To publish them for everyone, update this notebook and redeploy.
