## Nested Conditionals

This program checks whether a student is ready for school and sports practice. Each inner `if` runs only when every enclosing condition is true. Python indentation shows which `else` belongs to each `if`.

Edit the four Boolean values in the runner below, then click **Run** or press **Ctrl+Enter / Cmd+Enter**. Keep the language set to **Python**. **Save** stores edits in this browser; **Clear Storage** restores the starter.

With the starter values, the expected output is:

> Logged in, but homework still needs to be finished.

Test all five branches: set `logged_in` to `False`; leave `homework_finished` as `False`; finish homework but leave lunch unpacked; pack lunch without practice today; then set all four values to `True`. Predict the result before each run.

Python runs directly in your browser. The first run downloads the Python runtime and may take a moment; later runs reuse it. An internet connection is required to load the runtime.


```python
# CODE_RUNNER: Change the student Boolean values, predict the output, then press Run to test each nested branch. | python_runtime: browser
student = {
    "logged_in": True,
    "homework_finished": False,
    "lunch_packed": True,
    "practice_today": True
}

# Complete nested conditional logic
if student["logged_in"]:
    if student["homework_finished"]:
        if student["lunch_packed"]:
            if student["practice_today"]:
                print("Ready for school and sports practice!")
            else:
                print("Ready for school! No practice today.")
        else:
            print("Almost ready, but need to pack lunch.")
    else:
        print("Logged in, but homework still needs to be finished.")
else:
    print("Student is not logged in. Please log in first.")
```
