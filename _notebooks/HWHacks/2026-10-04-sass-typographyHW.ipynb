## Refined HTML Submission

The cell below is the clean HTML submission. It uses semantic headings, paragraphs, an ordered list, and strong emphasis. Only the existing `ocs__container`, `ocs__card`, and `ocs__section-title` classes are needed; the shared Aesthetihawk SASS theme supplies the styling.

Edit the HTML in the **UI Runner**, then press **Run** or **Ctrl+Enter / Cmd+Enter** to render it. Changes appear only when you run the code. **Save** keeps edits in this browser; **Clear Storage** restores the submitted HTML. Move the pointer over the card to see the theme's highlight.

```python
%%html
<!-- UI_RUNNER: Typography Homework: semantic HTML with the shared OCS theme | editor: html -->
<div class="ocs__container">
  <article class="ocs__card">
    <h2 class="ocs__section-title">How to Pat a Cat</h2>
    <h3 class="ocs__section-title">Have you ever wanted to...</h3>
    <p>Pat a cat? Well, good for you! We will give <strong>step-by-step instructions</strong> on how to pat a cat!</p>
    <h4 class="ocs__section-title">How to pat a cat:</h4>
    <ol>
      <li>Put out your hand.</li>
      <li>Touch the cat.</li>
      <li>Move your hand back and forth.</li>
    </ol>
    <h3 class="ocs__section-title">Congrats! You pat a cat :D</h3>
  </article>
</div>
```
