## Popcorn Hack 2

```python
# CODE_RUNNER: Decomposition & Helper Procedures. | python_runtime: browser
def is_priority(incident_type):
    return incident_type == "Medical Emergency"

def process_report(location, incident_type):
    priority = is_priority(incident_type)
    if priority:
        print(f"PRIORITY: {incident_type} reported in {location}")
    else:
        print(f"STANDARD REPORT: {incident_type} reported in {location}")

# Test calls matching the provided pseudocode
process_report("Poway", "Medical Emergency")
process_report("Poway", "Power Outage")
```

## Python Calling Procedures

```python
# CODE_RUNNER: Run the emergency-response example and compare priority and routine incidents. | python_runtime: browser
# Simplified emergency-response program

def check_priority(incident_type):
    return incident_type in ("fire", "medical emergency")


def dispatch_response(location, incident_type):
    priority = check_priority(incident_type)
    if priority:
        print(f"Priority response: dispatch a team to {location} for {incident_type}.")
    else:
        print(f"Routine response: schedule assistance at {location} for {incident_type}.")


dispatch_response("Community Center", "fire")
dispatch_response("City Park", "medical emergency")
dispatch_response("Library", "fallen branch")
```
