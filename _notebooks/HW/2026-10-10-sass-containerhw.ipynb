## Popcorn Hack Challenge

The parade invitation uses one `ocs__container` and an `ocs__card`, with a section
title, a short description, and three schedule cells. The start time uses the
accent modifier. All styling comes from the shared OCS SASS classes.

Press **Run** to render the invitation. Resize the window to check the grid:
three columns on desktop, two on tablet, and one on mobile (480px or narrower).


Homework Hack
