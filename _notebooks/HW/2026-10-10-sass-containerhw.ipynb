## Popcorn Hack Challenge

The parade invitation uses one `ocs__container` and an `ocs__card`, with a section
title, a short description, and three schedule cells. The start time uses the
accent modifier. All styling comes from the shared OCS SASS classes.

Press **Run** to render the invitation. Resize the window to check the grid:
three columns on desktop, two on tablet, and one on mobile (480px or narrower).


## Homework Hack Challenge

Render the Poway Rotary Parade information using the shared OCS theme: a container and card, a section title and description, a schedule grid, an event-status table, and a callout.

Press **Run** (or **Ctrl+Enter / Cmd+Enter**) to render the completed code below. Edit the HTML inside the JavaScript template string and press **Run** again to see your changes. Use only the OCS classes for styling, and resize the page to check the schedule layout.
