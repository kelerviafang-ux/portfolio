## Popcorn Hack 1

An intercepted transmission has a real message hidden inside decoy characters at fixed positions. Use only indexing and slicing — no string methods yet — to pull the real words out.

The completed code extracts `GHOST`, `THE EAGLE LANDS AT DAWN`, and `87` from a padded transmission. It also uses indexing to get the first and last letters of the agent name. Slice end positions are excluded.

Each Popcorn Hack runs independently. Click **Run** (or press **Ctrl+Enter / Cmd+Enter**) in its browser Python runner.


```python
# CODE_RUNNER: Popcorn Hack 1 - Decode fixed positions using indexing and slicing only. | python_runtime: browser

raw_transmission = "@@GHOST##THE EAGLE LANDS AT DAWN!!87$$"

# [2:7] includes indexes 2 through 6, skipping the @@ and ## decoys.
agent = raw_transmission[2:7]
message = raw_transmission[9:32]
confidence_str = raw_transmission[34:36]

# A single index selects one character; -1 selects the last character.
first_letter = agent[0]
last_letter = agent[-1]

print("Agent:", agent)
print("Message:", message)
print("Confidence:", confidence_str)
print("First letter:", first_letter)
print("Last letter:", last_letter)
```

## Popcorn Hack 2

This transmission is full of noise — stray punctuation, extra whitespace, and mismatched case. Clean it up and search it using string methods.

The completed code uses `.strip()`, `.replace()`, `.upper()`, `.lower()`, `.split()`, `.find()`, and `.count()`. It recovers `GHOST: THE EAGLE LANDS AT DAWN`, searches for `dawn` without regard to case, and counts the original exclamation marks.


```python
# CODE_RUNNER: Popcorn Hack 2 - Clean and search a noisy transmission with string methods. | python_runtime: browser

raw_transmission = "   gHoSt: tHe-EaGlE-LaNdS-aT-dAwN!!!  "

# Each method returns a new string, so save the result rather than changing raw.
clean = raw_transmission.strip().replace("!", "").replace("-", " ").upper()
print("Clean transmission:", clean)
print("Recovered fragments:", clean.split(": "))

# find() returns -1 when the word is missing; index zero is also a valid match.
dawn_index = clean.lower().find("dawn")
found_dawn = dawn_index >= 0
print("Found 'dawn':", found_dawn)
print("Dawn index:", dawn_index)
print("Noise marks removed:", raw_transmission.count("!"))
```

## Popcorn Hack 3

Combine indexing/slicing, string methods, and formatting to decode a CSV-style transmission. Extract the four fields, check the year with a slice, clean the message, and search for `dawn`.

Then create an intercept report using concatenation, `str()`, an f-string, and `.join()`, followed by a Markdown table row and a JSON line. JSON booleans must be lowercase `true` or `false`.


```python
# CODE_RUNNER: Popcorn Hack 3 - Decode CSV and build formatted intercept reports. | python_runtime: browser, literal: true

raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

# Clean the outside before splitting the CSV into four fields.
segments = raw_transmission.strip().split(",")
print("Segments:", segments)

# Indexes start at zero; the confidence field is the fourth entry.
date = segments[0]
agent = segments[1]
message_code = segments[2]
confidence_str = segments[3]
print("Date:", date, "| Agent:", agent, "| Message code:", message_code, "| Confidence:", confidence_str)

year_check = date[:4] == "2026"
print("Year check:", year_check)
message = message_code.replace("-", " ")
found_dawn = "dawn" in message.lower()
print("Found 'dawn':", found_dawn)

# Concatenation combines strings; str() lets a numeric value join the report.
confidence = int(confidence_str)
report = agent + " reports: " + message
report_with_number = report + " (confidence: " + str(confidence) + "%)"
print("Concatenation report:", report_with_number)

# An f-string mixes strings and numbers directly; join() combines a list of words.
report_f = f"{date}: {agent} reports {message} ({confidence}% confidence)"
print("F-string report:", report_f)
fragments = message.split()
print("Joined message:", " ".join(fragments))

markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
print("Markdown row:", markdown_row)

# Double braces escape JSON braces; lower() produces a JSON Boolean.
json_line = (
    f'{{"date": "{date}", "agent": "{agent}", "message": "{message}", '
    f'"confidence": {confidence_str}, "contains_dawn": {str(found_dawn).lower()}}}'
)
print("JSON line:", json_line)
```

## HW Hack

```python
# CODE_RUNNER: Homework - Crack the Final Transmission using slicing, methods, and formatting | python_runtime: browser, literal: true

raw_transmission = "  2026-09-15,GHOST,THE-EAGLE-LANDS-AT-DAWN,87  "

# strip() cleans the outside of the row; split(",") makes a four-item list.
segments = raw_transmission.strip().split(",")
print("Segments:", segments)

# List indexes start at zero, so the fourth field is at index 3.
date = segments[0]
agent = segments[1]
message_code = segments[2]
confidence_str = segments[3]
print("Date:", date, "| Agent:", agent, "| Message code:", message_code, "| Confidence:", confidence_str)

# The end index is excluded: [:4] takes exactly the first four characters.
year_check = date[:4] == "2026"
print("Year check:", year_check)

# replace() returns a new string with spaces in place of the message dashes.
message = message_code.replace("-", " ")

# lower() makes the search case-insensitive; IN returns a Boolean.
found_dawn = "dawn" in message.lower()
print("Found 'dawn':", found_dawn)

# An f-string inserts values, while the percent sign and table separators stay literal.
markdown_row = f"| {date} | {agent} | {message} | {confidence_str}% |"
print("Markdown row:", markdown_row)

# Double braces produce literal JSON braces inside an f-string.
# The confidence is unquoted, and str(...).lower() produces JSON's true/false.
json_line = (
    f'{{"date": "{date}", "agent": "{agent}", "message": "{message}", '
    f'"confidence": {confidence_str}, "contains_dawn": {str(found_dawn).lower()}}}'
)
print("JSON line:", json_line)
```
